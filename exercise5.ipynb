```python
# Sistem Pengelolaan Nilai Siswa
siswa = [
    {"nama": "Evelyn", "nilai": 88},
    {"nama": "Budi", "nilai": 75},
    {"nama": "Citra", "nilai": 92},
    {"nama": "Doni", "nilai": 60},
    {"nama": "Ario", "nilai": 80},
]

# Fungsi Tampilkan Daftar Nilai
def tampilkan_nilai(siswa):
    print("\nDaftar Nilai Siswa:")
    for s in siswa:
        print(f"- {s["nama"]}: {s["nilai"]}")
tampilkan_nilai(siswa)

# Bubble sort berdasarkan nilai
```
