```python
'''
Sintax kelas di python:

class NamaKelas:
    isi dark kelas
    # mengisi atribut/data dari kelas
    def__init__(self, attribut1, attribut2, dst)
      # self untuk mengarah ke kelas itu sendiri
      self.atribut1 = atribut1
      dst

    # method di dalam kelas
    def nama_method():
        isi method

sintax membuat object:

nama_object = NamaKelas(isiatribut1 dst)
nama_object.nama_method atau nama atribut
'''

# nama kelas
class Mobil:
    def __init__(self, merk, warna, tahun):
        self.merk = merk # atribut/data mark
        self.warna = warna # atribut/data warna
        self.tahun = tahun # atribut/data tahun
    
    def jalan(self):
        print (f"mobil {self.merk} sedang berjalan")

    def berhenti(self):
        print (f"mobil {self.merk} berhenti")

    def klakson(self):
        print (f"mobil {self.warna} klakson")

# membuat object dari kelas
mobil1 = Mobil("Toyota", "Hijau", 2020) # set atribut
mobil2 = Mobil("Mitsubhisi", "Putih", 2020)
mobil3 = Mobil("Honda", "Hitam", 2025)

# memanggil method dari kelas
mobil1.jalan()
mobil2.jalan()
mobil3.jalan()

mobil1.berhenti()
mobil2.berhenti()
mobil3.berhenti()

mobil1.klakson()
mobil2.klakson()
mobil3.klakson()

# kita panggil atribut/data
print(mobil1.warna)
print(mobil3.tahun)
mobil1.tahun = 2023
print (mobil1.tahun)
# atribut itu ga pakai ()
```

```python
# Encapsulation = mmenyembunyikan data / atribut agar tidak bisa diakses
# biasanya digunakan untuk melindungi data

class AkunBank:
    def __init__(self, nama, saldo):
        self.nama = nama # publikc atribut
        self.__saldo = saldo # private saldo atribut
        # perhatikan penggunaan __

    # setter dan getter
    def setSaldo(self, saldo):
        self.__saldo = saldo

    def getSaldo(self):
        print (f"Saldo kamu {self.__saldo}")

# membuat object/akun
akun1 = AkunBank("Hafiz", 7000)
print (akun1.nama)
akun1.nama = "Anton"
print (akun1.nama)
akun1.getSaldo()
akun1.setSaldo(9000)
akun1.getSaldo()
```

```python
# inheritance (pewarisan)
# -> sebuah kelas bisa mewarisi atribut dan method dari class lain
# mirip seperti anak mewarisi sifat dari orang tua 

# parent kelas/super class
class Hewan: 
    # atributnya
    def __init__(self, nama):
        self.nama = nama

    def suara(self):
        print ("Hewan Bersuara")

    def jumlah_kaki(self):
        print ("Hewan berkaki 4")
# child kelas
class Kucing(Hewan): # kelas kucing mewarisi dari kelas hewan
    def suara(self):
        print("Meong-meong")

class Anjing(Hewan): # kelas anjing mewarisi dari kelas hewan
    def suara(self):
        print ("Guk-guk")

# buat object
kucing1 = Kucing("mimi") # isi kurung adalah atributnya
anjing1 = Anjing("leki")
hewan1 = Hewan("lala")
kucing1.suara()
print(kucing1.nama)
anjing1.suara()
hewan1.suara()
anjing1.jumlah_kaki()
        
```

```python
# polimorfisme
# objek berbeda bisa memiliki method dengan nama sama, tapi implementasinya berbeda.

class Person:
    def __init__(self, nama, usia):
        self.nama = nama
        self.usia = usia

    def greet(self):
        print("Halo teman-teman")

class English(Person):
    def greet(self): # ovveride
        print("Halo friends")

class German(Person):
    def greet(self):
        print("tschuss preunds")

# buat object
person1 = English("Hafiz", 20)
person2 = German("Gio", 21)

person1.greet() # nama method sama
person2.greet() # tapi implementasi beda
print (person2.usia)
```
